Handle Missing Values

```python
df.fillna(0, inplace=True)
df.dropna(inplace=True)
df.isnull().sum()
```

Check Duplicate Records

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

Statistical Summary

```python
df.describe()
```

```python
#Find Average Crop Yield
df["yield_kg_per_hectare"].mean()
```

```python
# Crop Type Analysis
df.groupby("crop_type")["yield_kg_per_hectare"].mean()
```

```python
#Region-wise Crop Yield
df.groupby("region")["yield_kg_per_hectare"].mean()
```

```python
#Irrigation Type Analysis
df.groupby("irrigation_type")["yield_kg_per_hectare"].mean()
```

```python
#Fertilizer Analysis
df.groupby("fertilizer_type")["yield_kg_per_hectare"].mean()
```

```python
#Find Top 10 Farms
top_farms = df.sort_values(
    by="yield_kg_per_hectare",
    ascending=False
)

top_farms.head(10)
```

Visualization

```python
#Soil Moisture vs Yield  Visualization
plt.figure(figsize=(8,5))

plt.scatter(
    df["soil_moisture_%"],
    df["yield_kg_per_hectare"]
)

plt.xlabel("Soil Moisture %")
plt.ylabel("Crop Yield")
plt.title("Soil Moisture vs Crop Yield")

plt.show()
```

```python
#Rainfall Analysis
plt.figure(figsize=(8,5))

plt.scatter(
    df["rainfall_mm"],
    df["yield_kg_per_hectare"]
)

plt.xlabel("Rainfall")
plt.ylabel("Crop Yield")
plt.title("Rainfall vs Crop Yield")

plt.show()
```

```python
#Crop Yield by Crop Type
crop_yield = df.groupby("crop_type")[
    "yield_kg_per_hectare"
].mean()

crop_yield.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Average Yield by Crop Type")
plt.ylabel("Yield")

plt.show()
```

Correlation Heatmap

```python
plt.figure(figsize=(12,8))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
